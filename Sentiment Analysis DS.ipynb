```python
# sample data - 500
# remove - html
# remove special 
# lowercase
# remove stopwords
# stemming
```

```python
data = data.sample(500)
```

```python
data
```

```python
data['sentiment'].unique()
```

```python
data['sentiment'].replace({'negative':0, 'positive':1},inplace=True)
```

```python
data
```

```python
import re
```

```python
def clean_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean,'',text)
```

```python
data['review']=data['review'].apply(clean_html)
```

```python
data
```

```python
def remove_special(text):
    x = ''
    for i in text:
        if i.isalnum():
            x += i
        else:
            x += ' '
    return x
```

```python
data['review']=data['review'].apply(remove_special)
```

```python
data
```

```python
data['review'].iloc[5]
```

```python
def covert_low(text):
    return text.lower()
```

```python
data['review']=data['review'].apply(covert_low)
```

```python
data['review'].iloc[5]
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
a = stopwords.words('english')
```

```python
a
```

```python
def remove_stop(text):
    x = []
    for i in text.split():
        if i not in a:
            x.append(i)
        else:
            pass
    return x
```

```python
data['review']=data['review'].apply(remove_stop)
```

```python
data['review'].iloc[5]
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
from nltk.stem import PorterStemmer

def stem_words(text):
    ps = PorterStemmer()  # Initialize the stemmer
    x = []
    for i in text:
        x.append(ps.stem(i))
    return x  # Return the result after the loop finishes
```

```python
data['review']=data['review'].apply(stem_words)
```

```python
data
```

```python
def join (list_input):
    return ' '.join(list_input)
```

```python
data['review'].iloc[5]
```

```python
data['review']=data['review'].apply(join)
```
